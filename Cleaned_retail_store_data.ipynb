```python
missing_price = (
    df["Price Per Unit"].isnull()
    & df["Quantity"].notnull()
    & df["Total Spent"].notnull()
)

df.loc[missing_price, "Calculated Price"] = (
    df.loc[missing_price, "Total Spent"]
    / df.loc[missing_price, "Quantity"]
)

print(
    df.loc[
        missing_price,
        ["Category", "Quantity", "Total Spent", "Calculated Price"]
    ].head(10)
)
```

```python
missing_item_with_price = (
    df["Item Cleaned"].isnull()
    & df["Calculated Price"].notnull()
)

calculated_matches = pd.MultiIndex.from_arrays([
    df.loc[missing_item_with_price, "Category"],
    df.loc[missing_item_with_price, "Calculated Price"]
]).map(item_lookup)

df.loc[missing_item_with_price, "Item Cleaned"] = calculated_matches

print("Missing items remaining:", df["Item Cleaned"].isnull().sum())
```

```python

import pandas as pd

# Ensure the date column has the correct format
df["Transaction Date"] = pd.to_datetime(
    df["Transaction Date"], errors="coerce"
)

# Create month and revenue columns
df["Month"] = (
    df["Transaction Date"]
    .dt.to_period("M")
    .astype("string")
)

df["Revenue"] = (
    df["Price Cleaned"] * df["Quantity"]
)

# Create a final dataset using cleaned values
df_final = pd.DataFrame({
    "Transaction ID": df["Transaction ID"],
    "Customer ID": df["Customer ID"],
    "Category": df["Category"],
    "Item": df["Item Cleaned"],
    "Price Per Unit": df["Price Cleaned"],
    "Quantity": df["Quantity"],
    "Total Spent": df["Total Spent"],
    "Revenue": df["Revenue"],
    "Payment Method": df["Payment Method"],
    "Location": df["Location"],
    "Transaction Date": df["Transaction Date"],
    "Month": df["Month"],
    "Discount Applied": df["Discount Applied"]
})

print("Final dataset shape:", df_final.shape)
df_final.head()
```

```python

# Save the final dataset as a CSV file
output_file = "Retail_Store_Sales_Cleaned.csv"

df_final.to_csv(output_file, index=False)

print("File saved successfully:", output_file)
print("Rows exported:", len(df_final))
print("Columns exported:", len(df_final.columns))

# Download the CSV from Google Colab
from google.colab import files
files.download(output_file)
```
